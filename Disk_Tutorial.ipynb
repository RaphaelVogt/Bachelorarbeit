```python
import torch

from embedding.poincare import PoincareDisk
```

```python
# create Poincare Disk model from existing data points
data = torch.tensor([-0.5 - 0.5j, -0.5 + 0.5j, 0.5 - 0.5j, 0.5 + 0.5j])
D1 = PoincareDisk()
D1.from_tensor(data)
print(D1.data_points)
D1.visualize(size=3, circle=True)
```

```python
# create Poincare Disk model by sampling random points
D2 = PoincareDisk()
D2.sample_random(num_points=300, radius=1)
D2.visualize(size=3, circle=True)
```

```python
# compare distance computations in D1
print(f"Direct distance between point 0 and point 2 in D1 is {D1.riemann_distance(0, 2).item()}")
D1.moebius(0, 2)
print(f"Distance between point 0 and point 2 in D1 using moebius transform is {D1.distance_center(0).item()}")
D1.visualize(size=3, circle=True)
```
